```python
#### Start of "activity" filtering ####

df_clean["activity"] = df_clean["activity"].str.strip().str.title()
df_clean["activity"] = df_clean["activity"].fillna("Unknown")

def clean_activity(activity):
    for value in ["Swimming", "Diving","Surfing", "Spearfishing", "Wading", "Fishing", "Rescue", "Snorkeling" , "Hydrofoiling"]:
        if value.lower() in activity.lower():
            return value.lower()
        else:
            return "other"

df_clean["activity"].apply(clean_activity)

top_activities = df_clean["activity"].value_counts().head(10).index
```

```python
#Clean basic formatting, so "Australia" and "AUSTRALIA" count as one
df_clean["country"] = df_clean["country"].str.strip().str.title()

#Fix acronyms that .title() changes ("USA" -> "Usa")
df_clean["country"] = df_clean["country"].replace({"Usa": "USA", "Uae": "UAE"})

#The 10 most common countries
top_countries = df_clean["country"].value_counts().head(10).index

#Keep only rows with those countries
df_clean_top_countries = df_clean[df_clean["country"].isin(top_countries)]

df_clean_top_countries["country"].value_counts()
```

```python
#Clean basic formatting, so "Florida" and "FLORIDA " count as one
df_clean["state"] = df_clean["state"].str.strip().str.title()

#The 10 most common states
top_states = df_clean["state"].value_counts().head(10).index

#Keep only rows with those states
df_clean_top_states = df_clean[df_clean["state"].isin(top_states)]

df_clean_top_states["state"].value_counts()
```

```python
# This function will clean both Country and State
def get_top_values(df, column, n=10):
    """Return the rows whose value in `column` is among the n most common."""
    top_values = df[column].value_counts().head(n).index
    return df[df[column].isin(top_values)]

df_clean_top_countries = get_top_values(df_clean, "country")
df_clean_top_states = get_top_values(df_clean, "state")
```

```python
import re

def identify_period(date):
    if re.match(r"\d{4}-\d{4}", date):
        return "Period"
    else:
        return date

df_clean["date"].apply(identify_period)
df_clean[ df_clean["date"] != "Period" ]
```

```python
#Find the month name in the date text (e.g., "18th September" -> "Sep")
df_clean["month"] = (
    df_clean["date"]
    .astype(str)      #convert every value to text, so the regex works on all rows
    .str.lower()      #lowercase everything, so "Sep", "SEP" and "sep" all match the pattern
    .str.extract(r"(jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)")[0]  #find the first month abbreviation; [0] takes the matched text as a column
    .str.title()      #capitalise the result, so all months look the same ("Sep")
)

#Count incidents per month
df_clean["month"].value_counts()
```
